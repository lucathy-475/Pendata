## 3. Rumus Euclidean Distance
Rumus Euclidean Distance:
$$
d(x,y) = \sqrt{\sum_{i=1}^{n} (x_i - y_i)^2}
$$

Karena ada 8 kolom hasil one-hot encoding, maka rumusnya menjadi:
$$
d(x,y) = \sqrt{
(JK_{x1} - JK_{y1})^2 +
(JK_{x2} - JK_{y2})^2 +
(KF_{x1} - KF_{y1})^2 +
(KF_{x2} - KF_{y2})^2 +
(KF_{x3} - KF_{y3})^2 +
(KF_{x4} - KF_{y4})^2 +
(FB_{x1} - FB_{y1})^2 +
(FB_{x2} - FB_{y2})^2
}
$$

Keterangan:

- **JK** = atribut *Jenis_Kelamin*  
- **KF** = atribut *Kategori_Favorit*  
- **FB** = atribut *Frekuensi_Belanja*  
- **d(x,y)** = jarak antara data **x** dan data **y**

d(1,5) = \sqrt{6} \approx 2.45

## 7. Tabel Pengurutan Jarak

Setelah jarak antara setiap data dengan **Data ke-5 (target)** dihitung, langkah berikutnya adalah **mengurutkan jarak dari yang terkecil ke terbesar**.

## 8. Tentukan Nilai K

Misalkan digunakan:

$$
K = 3
$$

Maka **3 tetangga terdekat** adalah:

| Data | Umur |
|------|------|
| 2 | 37 |
| 3 | 39 |
| 1 | 44 |

## 9. Prediksi Usia

Karena **Usia** adalah atribut **numerik**, maka nilai prediksi dihitung menggunakan **rata-rata dari K tetangga terdekat**.

Tetangga terdekat yang diperoleh adalah:

- Data 2 → Usia = 37  
- Data 3 → Usia = 39  
- Data 1 → Usia = 44  

Sehingga perhitungan prediksi usia adalah:

$$
Usia_{prediksi} =
\frac{37 + 39 + 44}{3}
$$

$$
Usia_{prediksi} =
\frac{120}{3}
$$

$$
Usia_{prediksi} = 40
$$

## 10. Kesimpulan

Berdasarkan perhitungan menggunakan metode **K-Nearest Neighbor (KNN)** dengan nilai **K = 3**, diperoleh tiga tetangga terdekat dari data target (Data ke-5), yaitu **Data 2, Data 3, dan Data 1** dengan jarak masing-masing **2.000, 2.000, dan 2.449**.

Karena atribut yang diprediksi yaitu **Usia** merupakan atribut **numerik**, maka nilai prediksi dihitung menggunakan **rata-rata usia dari tiga tetangga terdekat**.

Hasil perhitungan diperoleh:

$$
Usia_{prediksi} = 40
$$

Sehingga dapat disimpulkan bahwa **perkiraan usia untuk Data ke-5 berdasarkan metode KNN adalah sekitar 40 tahun**.
