# Data Customer

## Deskripsi Dataset Pelanggan

Dataset ini berisi data pelanggan yang digunakan untuk merepresentasikan karakteristik demografi, perilaku belanja, serta preferensi pelanggan dalam suatu periode tertentu. Setiap baris dalam dataset merepresentasikan satu pelanggan, sedangkan setiap kolom berisi atribut yang dapat digunakan untuk analisis deskriptif, eksplorasi data, maupun pemodelan sederhana.

Berdasarkan file dataset yang digunakan, data ini terdiri dari 180 pelanggan dengan 13 atribut. Dataset ini memadukan variabel numerik dan variabel kategorikal, sehingga cocok untuk berbagai jenis analisis data seperti analisis perilaku pelanggan, segmentasi pasar, maupun eksplorasi pola belanja.

Beberapa atribut utama yang terdapat dalam dataset ini antara lain:

- ID pelanggan sebagai identitas unik
- Umur pelanggan
- Penghasilan pelanggan
- Total belanja bulanan
- Jumlah transaksi per bulan
- Jenis kelamin pelanggan
- Kota tempat tinggal pelanggan
- Kategori produk favorit
- Tingkat pendidikan pelanggan
- Frekuensi belanja
- Tingkat kepuasan pelanggan
- Status keanggotaan (member)
- Penggunaan promo

Dataset ini memungkinkan berbagai jenis analisis, seperti:

- Analisis distribusi umur pelanggan
- Analisis penghasilan dan pola belanja
- Analisis frekuensi transaksi pelanggan
- Segmentasi pelanggan berdasarkan preferensi dan kepuasan
- Analisis penggunaan promo dan status member terhadap perilaku belanja

Dataset ini juga tidak memiliki nilai kosong sehingga siap digunakan untuk analisis lebih lanjut.

## Deskripsi Atribut

Berikut merupakan atribut/fitur yang terdapat dalam dataset pelanggan:

| No | Atribut | Deskripsi |
|----|--------|-----------|
| 1 | Customer_ID | Nomor identitas unik pelanggan |
| 2 | Umur | Usia pelanggan dalam tahun |
| 3 | Penghasilan_Juta | Penghasilan pelanggan dalam juta rupiah |
| 4 | Total_Belanja_Bulanan_Juta | Total pengeluaran belanja pelanggan per bulan (juta rupiah) |
| 5 | Transaksi_Per_Bulan | Jumlah transaksi yang dilakukan pelanggan dalam satu bulan |
| 6 | Jenis_Kelamin | Jenis kelamin pelanggan (Laki-laki / Perempuan) |
| 7 | Kota | Kota tempat tinggal pelanggan |
| 8 | Kategori_Favorit | Kategori produk yang paling sering dibeli pelanggan |
| 9 | Pendidikan | Tingkat pendidikan terakhir pelanggan |
| 10 | Frekuensi_Belanja | Kategori frekuensi belanja pelanggan |
| 11 | Tingkat_Kepuasan | Tingkat kepuasan pelanggan terhadap layanan atau produk |
| 12 | Status_Member | Status keanggotaan pelanggan (0 = Tidak Member, 1 = Member) |
| 13 | Menggunakan_Promo | Status penggunaan promo saat berbelanja (0 = Tidak, 1 = Ya) |

## Missing Value

Berdasarkan pemeriksaan data pada seluruh atribut dalam dataset pelanggan, tidak ditemukan nilai kosong (missing value).

Setiap kolom memiliki jumlah data yang lengkap sesuai dengan jumlah baris dataset, yaitu sebanyak 180 data pelanggan. Hal ini menunjukkan bahwa dataset berada dalam kondisi lengkap dan tidak memerlukan proses penanganan missing value seperti imputasi ataupun penghapusan data.

Dengan demikian, dataset ini dapat langsung digunakan untuk proses analisis data maupun pemodelan lebih lanjut.

## Outlier

Pemeriksaan awal terhadap variabel numerik dalam dataset menunjukkan bahwa nilai data masih berada dalam rentang yang wajar.

Beberapa variabel numerik yang diperiksa antara lain:

- **Umur** berada pada rentang 17–55 tahun, yang masih realistis untuk data pelanggan.
- **Penghasilan_Juta** berada dalam kisaran penghasilan yang masih wajar untuk populasi pelanggan.
- **Total_Belanja_Bulanan_Juta** menunjukkan variasi pengeluaran yang masih masuk akal.
- **Transaksi_Per_Bulan** berada pada rentang 0–8 transaksi per bulan yang masih tergolong normal dalam aktivitas belanja.

Berdasarkan pemeriksaan awal tersebut, dataset ini **tidak menunjukkan adanya outlier yang signifikan** pada variabel numerik.
