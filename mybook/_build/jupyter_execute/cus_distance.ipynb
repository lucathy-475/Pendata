# Distance in Customer Data

## Mengukur Jarak pada Tipe Data Campuran

Dalam analisis data, sering dijumpai dataset yang memiliki tipe atribut campuran, yaitu atribut numerik dan kategorikal. Untuk menghitung jarak pada data dengan tipe campuran seperti ini, dapat digunakan pendekatan **Gower Distance**, yaitu metode yang menghitung jarak dengan menyesuaikan tipe atribut yang digunakan.

Dataset Customer yang digunakan terdiri dari beberapa atribut seperti:

- Umur
- Penghasilan_Juta
- Total_Belanja_Bulanan_Juta
- Transaksi_Per_Bulan
- Jenis_Kelamin
- Kota
- Kategori_Favorit
- Pendidikan
- Frekuensi_Belanja
- Tingkat_Kepuasan
- Status_Member
- Menggunakan_Promo

Setiap atribut akan dihitung jaraknya berdasarkan tipe datanya.

# Atribut Numerik

Untuk atribut numerik, jarak dihitung menggunakan normalisasi berdasarkan range.

Rumus:

$$
d_{ij} = \frac{|x_i - x_j|}{\text{range}}
$$

Contoh perhitungan atribut **Umur**.

Range umur:

$$
\text{range} = 55 - 17 = 38
$$

Perhitungan jarak:

$$
d = \frac{|25 - 30|}{38}
$$

$$
= \frac{5}{38}
$$

$$
\approx 0.132
$$

---

# Atribut Nominal

Untuk atribut nominal digunakan metode **Simple Matching**.

Rumus:

$$
d =
\begin{cases}
0, & \text{jika nilai sama} \\
1, & \text{jika nilai berbeda}
\end{cases}
$$

Contoh atribut **Kota**.

Customer 1 = Surabaya  
Customer 2 = Jakarta  

Karena berbeda, maka:

$$
d = 1
$$

---

# Atribut Binary

Untuk atribut binary, jarak dihitung dengan metode yang sama seperti nominal.

Contoh atribut **Status_Member**.

Customer 1 = Ya  
Customer 2 = Tidak  

Karena berbeda, maka:

$$
d = 1
$$

---

# Atribut Ordinal

Untuk atribut ordinal, nilai kategori diubah menjadi ranking terlebih dahulu.

Contoh atribut **Frekuensi_Belanja**:

| Kategori | Ranking |
|------|------|
| Jarang | 1 |
| Sedang | 2 |
| Sering | 3 |

Normalisasi:

$$
z = \frac{r-1}{m-1}
$$

Misalkan:

Customer 1 = Sedang → r = 2  
Customer 2 = Sering → r = 3  

$$
z_1 = \frac{2-1}{3-1} = 0.5
$$

$$
z_2 = \frac{3-1}{3-1} = 1
$$

Jarak:

$$
d = |0.5 - 1|
$$

$$
= 0.5
$$

---

# Perhitungan Total Distance

Setelah semua atribut dihitung jaraknya, maka nilai distance dijumlahkan dan dirata-ratakan.

Rumus:

$$
D(i,j) = \frac{\sum d_k}{p}
$$

dimana:

- $d_k$ = jarak pada atribut ke-k
- $p$ = jumlah atribut

Nilai distance yang lebih kecil menunjukkan bahwa dua customer memiliki tingkat kemiripan yang lebih tinggi.

---

# Interpretasi

Nilai distance digunakan untuk mengetahui tingkat kemiripan antar customer.

- Distance kecil → customer sangat mirip
- Distance besar → customer semakin berbeda

Metode ini dapat digunakan untuk berbagai analisis seperti:

- Customer Segmentation
- Clustering
- Recommendation System
