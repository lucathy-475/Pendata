# Visualisasi Distribusi Data

Visualisasi distribusi data adalah teknik grafis untuk menunjukkan bagaimana nilai-nilai dalam kumpulan data tersebar, mengelompok, atau membentuk pola tertentu (seperti Histogram, Boxplot, Density plot, Scatter Plot, Pair Plot dll).

# Scatter Plot

Scatter Plot digunakan untuk melihat hubungan antara dua variabel numerik, yaitu petal_length dan petal_width. Setiap titik merepresentasikan satu sampel bunga Iris, dan warna menunjukkan kelas (species).

Berdasarkan visualisasi, terlihat bahwa:

- Iris-setosa terpisah sangat jelas dari dua spesies lainnya.

- Iris-versicolor dan Iris-virginica memiliki sedikit overlap, tetapi masih dapat dibedakan.

- Terdapat hubungan positif antara petal_length dan petal_width (semakin panjang petal, semakin lebar).

Visualisasi ini menunjukkan bahwa fitur petal sangat efektif dalam membedakan spesies Iris.
