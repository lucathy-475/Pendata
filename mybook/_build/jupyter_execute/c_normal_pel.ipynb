# Contoh Normalisasi Data 

Dataset yang digunakan merupakan data pelanggan [`dataset_pelanggan.csv`](data/dataset_pelanggan.csv)

Normalisasi dilakukan pada atribut **Umur** dari dataset pelanggan.  
Tujuan normalisasi adalah menyamakan skala data agar tidak mempengaruhi proses analisis data mining.

## Statistik Data Umur

| Parameter | Nilai |
|---|---|
| Jumlah Data | 180 |
| Umur Minimum (Xmin) | 17 |
| Umur Maksimum (Xmax) | 55 |
| Mean (μ) | 37.53 |
| Standar Deviasi (σ) | 10.86 |

Rentang data:

$$
X_{max} - X_{min} = 55 - 17 = 38
$$

---

## 1. Min-Max Normalization

Metode ini mengubah nilai data ke rentang **0 sampai 1**.

## Rumus

$$
X' = \frac{X - X_{min}}{X_{max} - X_{min}}
$$

## Rumus dengan Nilai Dataset

$$
X' = \frac{X - 17}{55 - 17}
$$

$$
X' = \frac{X - 17}{38}
$$

## Contoh Perhitungan

Jika umur pelanggan **44**:

$$
X' = \frac{44 - 17}{38}
$$

$$
X' = \frac{27}{38}
$$

$$
X' = 0.711
$$

---

## 2. Z-Score Normalization (Standardization)

Metode ini menormalisasi data berdasarkan **rata-rata dan standar deviasi**.

## Rumus

$$
X' = \frac{X - \mu}{\sigma}
$$

## Rumus dengan Nilai Dataset

$$
X' = \frac{X - 37.53}{10.86}
$$

## Contoh Perhitungan

Jika umur pelanggan **44**:

$$
X' = \frac{44 - 37.53}{10.86}
$$

$$
X' = \frac{6.47}{10.86}
$$

$$
X' = 0.596
$$

---

## 3. Decimal Scaling Normalization

Normalisasi dilakukan dengan memindahkan **titik desimal** berdasarkan jumlah digit maksimum.

Nilai maksimum umur adalah **55**, sehingga:

- jumlah digit = **2**
- maka **j = 2**

## Rumus

$$
X' = \frac{X}{10^j}
$$

## Rumus dengan Nilai Dataset

$$
X' = \frac{X}{10^2}
$$

$$
X' = \frac{X}{100}
$$

## Contoh Perhitungan

Jika umur pelanggan **44**:

$$
X' = \frac{44}{100}
$$

$$
X' = 0.44
$$

---

## 4. Mean Normalization

Mean normalization menggunakan **rata-rata dan rentang data**.

## Rumus

$$
X' = \frac{X - mean}{X_{max} - X_{min}}
$$

## Rumus dengan Nilai Dataset

$$
X' = \frac{X - 37.53}{55 - 17}
$$

$$
X' = \frac{X - 37.53}{38}
$$

## Contoh Perhitungan

Jika umur pelanggan **44**:

$$
X' = \frac{44 - 37.53}{38}
$$

$$
X' = \frac{6.47}{38}
$$

$$
X' = 0.170
$$

## Kesimpulan

Normalisasi pada atribut **Umur** dapat dilakukan dengan beberapa metode:

- **Min-Max Normalization**
- **Z-Score Normalization**
- **Decimal Scaling Normalization**
- **Mean Normalization**

Setiap metode memiliki cara perhitungan yang berbeda tetapi bertujuan sama yaitu **menyamakan skala data agar analisis data mining lebih akurat**.
