# Distance Measurement

## Pengertian

Mengukur jarak adalah proses menghitung tingkat kemiripan (similarity) atau ketidaksamaan (dissimilarity) antara dua objek data.
Konsep ini menjadi dasar dalam berbagai teknik data mining dan machine learning.

## Mengapa Mengukur Jarak Penting dalam Penambangan Data?

Dalam data mining, banyak algoritma bekerja berdasarkan konsep kedekatan antar data.
Jarak adalah cara kuantitatif untuk menentukan seberapa mirip atau berbeda dua data.

### 1. Dasar untuk Clustering (Pengelompokan)

Algoritma seperti:
- K-Means
- Hierarchical Clustering
- DBSCAN

mengelompokkan data berdasarkan jarak antar titik data.

Contoh:
Jika dua pelanggan memiliki usia dan pola belanja yang mirip (jaraknya kecil), maka mereka masuk cluster yang sama.

Tanpa pengukuran jarak → sistem tidak tahu mana data yang dekat atau jauh.

### 2. Digunakan dalam K-Nearest Neighbors (KNN)

Pada algoritma KNN => Data baru mengikuti kelas dari tetangga terdekatnya.
Jarak menentukan:
- Siapa tetangga terdekat
- Seberapa dekat mereka

Jika metrik jarak salah → hasil klasifikasi bisa salah.

### 3. Mengukur Kemiripan Data

Digunakan untuk mengetahui:

- Dokumen mana yang paling mirip
- Produk mana yang serupa
- Pasien mana yang memiliki kondisi hampir sama

Contoh metrik jarak:

- Euclidean Distance → data numerik
- Manhattan Distance → data multidimensi
- Cosine Similarity → data teks
- Hamming Distance → data biner

### 4. Deteksi Outlier (Anomali)

Data dengan jarak sangat jauh dari kelompok lain dapat dianggap sebagai anomali.

Contoh:
Dalam deteksi fraud, transaksi yang jauh dari pola normal dianggap mencurigakan.

### 5. Berpengaruh terhadap Akurasi Model

Pemilihan jenis jarak sangat krusial:
- Euclidean → numerik dengan skala sama
- Cosine → teks
- Hamming → biner

Jika salah memilih metrik:

- Cluster tidak masuk akal
- Prediksi bias
- Model gagal

## Konsep Dasar

Similarity → Nilai 0–1, semakin besar semakin mirip.
Dissimilarity (Distance) → Nilai minimum 0, semakin kecil semakin mirip.
Data Matrix → Representasi data dalam baris (objek) dan kolom (atribut).
Dissimilarity Matrix → Matriks jarak antar objek.

## Jenis Atribut

Nominal → Tanpa urutan
Ordinal → Ada urutan
Biner → Dua nilai (simetris & asimetris)
Numerik → Interval & rasio
Atribut campuran → Kombinasi beberapa tipe data

## Standarisasi Data

Z-score → Normalisasi menggunakan mean dan standar deviasi.
Mean Absolute Deviation → Lebih robust terhadap outlier.
Standarisasi penting agar atribut berskala besar tidak mendominasi perhitungan jarak.

## Kesimpulan

Mengukur jarak adalah fondasi dalam data mining karena hampir semua teknik berbasis pada konsep kedekatan atau kemiripan antar data.

Tanpa pengukuran jarak:

- Tidak ada clustering efektif
- Tidak ada KNN
- Tidak ada deteksi anomali berbasis jarak
- Tidak ada pengukuran kemiripan yang objektif
