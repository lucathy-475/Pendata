# Perhitungan WKNN menggunakan python

Notebook ini menunjukkan cara menghitung **missing value** menggunakan metode **Weighted K-Nearest Neighbor (WKNN) menggunakan python**.

Data sudah dinormalisasi menggunakan **Min-Max Normalization**.

## 1. Data Normalisasi

| No | IPK | PO | JML |
|---|---|---|---|
|1|0|0|0|
|2|0.5|0.5|1|
|3|1|0|0|
|4|0|0|1|
|5|0.5|0.5|0|
|6|1|1|1|
|7|0|0.5| ? |

Target data yang ingin diprediksi:

- IPK = 0
- PO = 0.5

## 2. Import Library

```python
import numpy as np
```

## 3. Data Training

```python
# fitur (IPK, PO)
X = np.array([
    [0,0],
    [0.5,0.5],
    [1,0],
    [0,0],
    [0.5,0.5],
    [1,1]
])

# label JML
y = np.array([0,1,0,1,0,1])
```

## 4. Data Target (Missing Value)

```python
target = np.array([0,0.5])
```

## 5. Menghitung Jarak Euclidean

```python
distances = np.sqrt(np.sum((X - target)**2, axis=1))
distances
```

## 6. Menghitung Bobot WKNN

```python
weights = 1 / distances
weights
```

## 7. Menghitung Prediksi (Weighted Average)

```python
prediction = np.sum(weights * y) / np.sum(weights)
prediction
```

## 8. Interpretasi Hasil

Nilai prediksi normalisasi sekitar **0.5**.

Mapping nilai:

| Normalisasi | Nilai JML |
|---|---|
|0|2|
|1|3|

Sehingga nilai JML diprediksi **≈ 2**.
