# KNN

K-Nearest Neighbor (KNN) adalah salah satu algoritma dalam penambangan data (data mining) dan machine learning yang digunakan untuk melakukan klasifikasi maupun prediksi (regresi) berdasarkan kedekatan atau kemiripan data.

Secara sederhana, KNN bekerja dengan cara membandingkan suatu data baru dengan sejumlah data yang paling dekat dengannya (tetangga terdekat) di dalam dataset. Kedekatan tersebut biasanya dihitung menggunakan ukuran jarak tertentu, seperti Euclidean Distance.

## Konsep KNN

- Menentukan nilai K
Nilai K adalah jumlah tetangga terdekat yang akan digunakan untuk menentukan hasil prediksi.

- Menghitung jarak
Sistem menghitung jarak antara data baru dengan seluruh data yang sudah ada di dataset.

- Menentukan tetangga terdekat
Dari semua data, dipilih K data dengan jarak paling dekat.

- Menentukan hasil
    - Pada klasifikasi: kelas yang paling banyak muncul dari tetangga tersebut menjadi hasil prediksi.
    - Pada regresi: nilai prediksi biasanya dihitung dari rata-rata nilai tetangga terdekat.

## Contoh Sederhana 

Misalnya ingin menentukan apakah suatu email termasuk spam atau tidak.
Algoritma KNN akan melihat beberapa email yang paling mirip dengan email baru tersebut. Jika sebagian besar email terdekat adalah spam, maka email baru juga diprediksi sebagai spam.

## Kelebihan dan Kekurangan 
### Kelebihan KNN
- Mudah dipahami dan diimplementasikan
- Tidak memerlukan proses pelatihan model yang kompleks
- Dapat digunakan untuk klasifikasi maupun regresi

### Kekurangan KNN

- Proses prediksi bisa lambat jika data sangat besar
- Sensitif terhadap pemilihan nilai K
- Dipengaruhi oleh skala dan distribusi data

## Kesimpulan
KNN adalah metode dalam penambangan data yang menentukan hasil klasifikasi atau prediksi dengan melihat kemiripan atau kedekatan suatu data dengan data lain yang sudah diketahui sebelumnya.
