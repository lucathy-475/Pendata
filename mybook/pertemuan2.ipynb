# Pertemuan 2
# Pengertian

Data Understanding adalah tahap memahami karakteristik data sebelum dilakukan analisis lebih lanjut. Tahap ini bertujuan untuk menemukan pola, membangun hipotesis awal, serta menghindari kesalahan dalam interpretasi data.

# Manfaat

- Membantu memilih teknik data mining yang tepat
- Meningkatkan akurasi model
- Mengurangi kesalahan interpretasi
- Menghasilkan analisis yang valid dan dapat ditindaklanjuti

# Komponen Utama

Pengumpulan data awal → Mengambil data dari sumber yang relevan.
Deskripsi data → Menjelaskan struktur data (jumlah baris, kolom, tipe data).
Eksplorasi Data (EDA) → Menganalisis pola, distribusi, dan hubungan antar variabel menggunakan statistik dan visualisasi.
Kualitas data → Memastikan data lengkap, konsisten, bebas duplikasi, dan minim kesalahan.

Jenis-Jenis Data

- Data terstruktur
- Data tidak terstruktur
- Data bahasa alami
- Data dari mesin
- Data audio, video, citra
- Data streaming
- Data berbasis graf

Konsep Penting

1. Korelasi → Mengukur hubungan antar dua variabel (-1 sampai +1).
2. Seleksi fitur → Memilih variabel paling berpengaruh untuk meningkatkan kinerja model.
3. Variabel independen (X) → Fitur/prediktor.
4. Variabel dependen (Y) → Target/label.

# Jenis Atribut

- Nominal (tanpa urutan)
- Ordinal (ada urutan, jarak tidak pasti)
- Biner (dua nilai: simetris & asimetris)
- Numerik (interval & rasio)

# Kesimpulan

Memahami data merupakan langkah fundamental dalam data mining agar analisis akurat, tidak menyesatkan, dan menghasilkan keputusan yang tepat.
